# Treating  null value in size coln


```python
df["Size"].unique()
```

```python
df.head()
```

```python
# from above table we know the size are given according to item puchased. so group by
df.groupby("Item Purchased")["Size"].unique()
```

```python
df.groupby("Category")["Size"].unique()
```

```python
#for filling coln: In electronicis not have size so relace it as Not Applicable.
#Treating null values or relacing all values.
df.loc[df["Category"]=="Electronics","Size"]="Not Applicable" 
```

```python
#to see changes
df.groupby("Category")["Size"].unique()
```

```python
#In Accessories not have size so relace it as Free Size.
df.loc[df["Category"]=="Accessories","Size"]="Free Size" 

# In footwear category size are not in L,M,S they are in number so replace it as Not Available.
df.loc[df["Category"]=="Footwear","Size"]="not available" 

#In clothing category treat null value we use mode (high number of freq: most common size)
# CODE EXPLAIN: *2 CONDn: Category = Clothing, Size is missing,
#               *Value to fill:most frequent size in Clothing
#              *0 return for if 2 similar mode then return firstly came.eg: L=100 tmis, S=100 times firstly L came then mode is L.
df.loc[
    (df["Category"] == "Clothing") & (df["Size"].isnull()),
    "Size"] = df[df["Category"] == "Clothing"]["Size"].mode()[0]
```

```python
#to see changes
df.groupby("Category")["Size"].unique()
```

```python
# to check treated null value for size coln
df.isnull().sum() #count of null value  
```

# Treating Review Rating null values

```python
df.head()
```

```python
#Mean of Review Rating(avg value)
df["Review Rating"].mean()
```

```python
#median of Review Rating (middle values)
df["Review Rating"].median()
```

```python
#mode of Review Rating(most freq value)
df["Review Rating"].mode()
```

```python
# review rating given according to item purchased so group by
#so we see each item purchased AVG rating
df.groupby("Item Purchased")["Review Rating"].mean()

```

```python
#filling null values of review rating with the mean of each item purchased.

#explain code:* Replaces only NaN values (in review rating coln fillna means null value)
#             *in that where each row gets the mean review rating of its item purchased. 

df["Review Rating"]=df["Review Rating"].fillna(df.groupby("Item Purchased")["Review Rating"].transform("mean"))
```

```python
# to check review rating null are handled
df.isnull().sum()
```

```python
df.head()
```

# Treating Previous Purchases null values

```python
# Previously Purchases so we treat null value as zero(0)
df["Previous Purchases"]=df["Previous Purchases"].fillna(0)
```

```python
#to check null are handled
df.isnull().sum()
```

# Treating Purchase Amount (USD) null values

```python
# calculates the average purchase amount (USD) for each item.
df.groupby("Item Purchased")["Purchase Amount (USD)"].mean()
```

```python
#filling null values of purchase amount with the mean of each item purchased.

df["Purchase Amount (USD)"]=df["Purchase Amount (USD)"].fillna(df.groupby("Item Purchased")["Purchase Amount (USD)"]
                                               .transform("mean"))
```

```python
# To check count of null
df.isnull().sum()
```
